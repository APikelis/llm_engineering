# Choosing a Model

## The Basics

Consider the following when putting together a candidate list of models to use
* Open vs. Closed Source - Licencing
* Chat / Reasoning /Hybrid
* Release date & Knowledge Cut Off
* Parameters - rule of thumb bigger is better
* Training Tokens
* Context Window - Bigger is also better but most otpimal phase is 40%

## Costs
* Inference Cost - API Charge + Runtime Compute
* Additonal Training Compute
* Build cost - for complementary features like RAG
* Time to Market - Using frontier could be faster but more expensive at runtime and vice versa
* Rate Limits - need to know especially for dealing with real-time/near-real time data
* Speed and Latency


# Laws & Benchmarks 
* Chinchilla Scaling Law - Roughly 20 training tokens per parameter for a compute-optimal training budget. 
    * Given a fixed training budget should you spend it on a larger model or more training tokens? What is the balance?

Other Laws: 
* Single Axis power laws ( learning curves )
* Joint additive model ( the common baseline )
* Compute scaling (loss vs training compute)

Remember Benchmarks are a starting point, they only give you eraly guodance to select a subset of models for evaluation

## Benchmark Limitations
* Training Data Contamination - Hard to keep test sets secret as info on these test sets get out and is involved in the training data of future models
    * Tests must get rotated to keep them fresh
* Not Consistently Applied - Hardware not specificied so some reproducibility wiggle room
* Narrow Scope - i.e., physics/Science Benchmark
* Hard to test nuanced reasoning
* Saturation - Older tests are now redundant as models are so much better
* Overfitting - models get trained to specifically be great at that test

LLMs may now be aware that they are being evaluated
