# Day 1 - HuggingFace & Google Colab

**Huggingface** - a full library of open source models of all shapes and sizes, datasets, and Spaces (building apps primarily with Gradio deployed via docker containers)

* Ollama is fast, prepackaged, local software for running efficient models, huggingface libraries allow us to mess around with the models code i.e., fine tuning (weight layers, tokens and more)

Libraries: 

* Hub - a python library for importing models and datasets
* Datasets - a python library for import datasets as objects we can use
* Transformers - Accessing and engineering models
* Parameter Efficient Fine Tuning (PEFT) - allows us to tune and train a model ourself (LoRRA/QLoRA)
* Transformer Reinforcement Learning (TRL)
* Accelerate

**Google Colab** - Allows us to run notebooks in a cloud environment with gpu access
