```python
system_prompt = """ You are a helpful assistant that analyses the content of a website and provides a short summery.
Ignore any text that may be nacigation related. You must repond in markdown format"""

user_prompt_prefix = """ Here are the contents of a website. Provide a short summary of this website in markdown. 
You must also summarise any news or announcements that may feature on the website"""

```

```python
# playing around with code
messages = [
    {"role": "system", "content": "You are an overly joyful assistant who always reponds in a verbose cheerful tone. You are also a bit of a know it all and will divulge obscure facts related to the users query"},
    {"role": "user", "content": "What is 2 + 2?"}
]

response = openai.chat.completions.create(model='gpt-4.1-nano', messages = messages)
response.choices[0].message.content
```

```python
# defining a function to create messages
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

# define a function to summarise a website - nesting in the messgaes_for function
def summarise(url):
    website = fetch_website_contents(url)
    messages = messages_for(website)
    response = openai.chat.completions.create(model='gpt-4.1-nano', messages=messages)
    return response.choices[0].message.content

# function to make the output readable
def display_summary(url):
    summary= summarise(url)
    display(Markdown(summary))

# test the function
display_summary("https://www.ibm.com/case-studies/mastercard-watson-advertising")


```

# Day 3 - Theory

3 Flavours of LLMs:

1. **Base Model** - Basic prediction of what comes next like in Auto-Complete. Previously you had to manually type out Q&A's followed by your conversation to simulate chat and force the model to predict an answer.

2. **Chat/Instruct Model** - Specialist model for conversation. The first instance of a system prompt that "describes the chat and the LLM's role in it" followed by User prompts and Assistant Replies.

3. **Reasoning/Thinking** - Models trained to output their thinking steps than an answers (CoT)

4. **Hybrid** - Combination of 3. & 4. however now effort is dynamic. It decides how much thinking is needed for the task.

DEF Budget Forcing - Forcing a model to reason longer (S1 Paper), e.g., Inserting the word wait into the thinking trace so next predicted words would be "Wait, am I sure?"

Key Points of Frontier Models
* Models are becoming increasingly knowledgable in specialised domains.
* They have limited knowledge beyond cut off dates (where tools, skills, and RAG come in handy)
* Can confidently make mistakes - They are prediction models at heart so must be grounded in reality.

## Illusion of Memory

Every time we call an API to interact with an LLM it is completely stateless - each call is indeoendent of each other. We can use basic trickery to fool the user in thinking there is memory - feed previous messages into a single call so that the LLM can use it as reference.


1. Every call to an LLM is stateless
2. We pass in the entire conversation so far in the input prompt, every time
3. This gives the illusion that the LLM has memory - it apparently keeps the context of the conversation
4. But this is a trick; it's a by-product of providing the entire conversation, every time
5. An LLM just predicts the most likely next tokens in the sequence

```python
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
load_dotenv(override=True)
openai_api_key = os.getenv("OPENAI_API_KEY")

if openai_api_key.startswith("sk-proj"):
    print("Valid API Key has been loaded")
```

```python

MODEL= "gpt-5-nano"
openai=OpenAI()
url='https://www.ibm.com/uk-en'
links=fetch_website_links(url)

# have gpt understand which links are relevant

link_system_prompt =""" You have been given a list of links found on a webpage.
You are able to decide which links are most relevant to include in a brochure about the company like links to an about page, Careers/Jobs or Company news.
You should respond in JSON like the example given.

Example:

{
    "links": [
        {"type": "about page", "url": "https://full.url/goes/here/about"},
        {"type": "careers page", "url": "https://another.full.url/careers"}
    ]
}
 """ 

def get_links_user_prompt(url):
    user_prompt = f""" 
    Here is the list of links on the website {url} - 
    please decide which of there a relevant web links for a brochure about the company, repond with the full url in JSON Format.
    Do not include Terms of Service, privacy, email links.

    Links (Some might be relative links)
    """
    links = fetch_website_links(url)
    user_prompt += "\n".join(links)

    link_counter = 0
    for link in links:
        link_counter = link_counter + 1

    user_prompt += f"\nNumber of links: {link_counter}"

    return user_prompt

def select_relevant_links(url):
    print(f"Selecting relevant links for {url} by calling GPT-5-nano")
    response = openai.chat.completions.create(
        model=MODEL,
        messages = [
            {"role":"system", "content":link_system_prompt},
            {"role": "user","content": get_links_user_prompt(url)}
        ],
        response_format={"type": "json_object"}
    )
    result = response.choices[0].message.content
    relevant_links = json.loads(result)
    print(f"Found {len(relevant_links['links'])} relevant links")
    return relevant_links

# select_relevant_links(url)
```

```python
# make the brochure

def fetch_page_and_relevant_links(url):
    contents = fetch_website_contents(url)
    relevant_links = select_relevant_links(url)
    result = f"## Landing Page: {url}\n## Relevant Links:\n"
    for link in relevant_links['links']:
        result += f"\n\n### Link: {link['type']}\n"
        result += fetch_website_contents(link['url'])
    return result

# print(fetch_page_and_relevant_links(url))

brochure_system_prompt = """
You are an assistant that analyzes the contents of several relevant pages from a company website
and creates a short brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks.
Include details of company culture, customers and careers/jobs if you have the information.
"""

def get_brochure_user_prompt(company_name, url):
    user_prompt = f""" 
    You are looking at website content for the company: {company_name}.
    Use this onformation to create a short brochure of the company in Markdown without code blocks. \n\n
    """
    user_prompt += fetch_page_and_relevant_links(url)
    user_prompt = user_prompt[:5_000] # trucnate as to not burn through tokens
    return user_prompt

def create_brochure(company_name, url):
    stream = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role":"system", "content": brochure_system_prompt},
            {"role": "user", "content": get_brochure_user_prompt(company_name, url)}
        ],
        stream = True
    )
    print("Creating brochure using GPT-4.1-mini")
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
    
```
